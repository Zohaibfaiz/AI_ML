```python
# ============================================================
# DECISION TREE CLASSIFICATION PROJECT
# PART 2 - MODEL BUILDING
# ============================================================

# ===============================
# 1. IMPORT REQUIRED LIBRARIES
# ===============================

from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

print("="*60)
print("Libraries Imported Successfully")
print("="*60)


# ===============================
# 2. FEATURE SELECTION
# ===============================

print("\nSelecting Features and Target...")

X = df.drop("target", axis=1)

y = df["target"]

print("\nFeature Shape :", X.shape)
print("Target Shape :", y.shape)


# ===============================
# 3. TRAIN TEST SPLIT
# ===============================

print("\nSplitting Dataset...")

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("\nTraining Samples :", len(X_train))
print("Testing Samples :", len(X_test))


# ============================================================
# MODEL 1
# GINI INDEX
# ============================================================

print("\n" + "="*60)
print("DECISION TREE (GINI)")
print("="*60)

gini_model = DecisionTreeClassifier(

    criterion="gini",

    max_depth=4,

    random_state=42

)

gini_model.fit(X_train, y_train)

print("Model Trained Successfully!")


# Prediction

gini_prediction = gini_model.predict(X_test)


# Accuracy

gini_accuracy = accuracy_score(
    y_test,
    gini_prediction
)

print("\nAccuracy :", round(gini_accuracy*100,2), "%")


# Confusion Matrix

print("\nConfusion Matrix")

print(
    confusion_matrix(
        y_test,
        gini_prediction
    )
)


# Classification Report

print("\nClassification Report")

print(
    classification_report(
        y_test,
        gini_prediction
    )
)


# ============================================================
# MODEL 2
# ENTROPY
# ============================================================

print("\n" + "="*60)
print("DECISION TREE (ENTROPY)")
print("="*60)

entropy_model = DecisionTreeClassifier(

    criterion="entropy",

    max_depth=4,

    random_state=42

)

entropy_model.fit(
    X_train,
    y_train
)

print("Model Trained Successfully!")


# Prediction

entropy_prediction = entropy_model.predict(
    X_test
)


# Accuracy

entropy_accuracy = accuracy_score(

    y_test,

    entropy_prediction

)

print("\nAccuracy :", round(entropy_accuracy*100,2), "%")


# Confusion Matrix

print("\nConfusion Matrix")

print(
    confusion_matrix(
        y_test,
        entropy_prediction
    )
)


# Classification Report

print("\nClassification Report")

print(
    classification_report(

        y_test,

        entropy_prediction

    )
)


# ============================================================
# ACCURACY COMPARISON
# ============================================================

print("\n" + "="*60)
print("MODEL COMPARISON")
print("="*60)

comparison = pd.DataFrame({

    "Model":[
        "Decision Tree (Gini)",
        "Decision Tree (Entropy)"
    ],

    "Accuracy":[
        gini_accuracy,
        entropy_accuracy
    ]

})

comparison["Accuracy"] = comparison["Accuracy"] * 100

print(comparison)


# ============================================================
# BAR CHART
# ============================================================

plt.figure(figsize=(6,5))

plt.bar(

    comparison["Model"],

    comparison["Accuracy"]

)

plt.ylabel("Accuracy (%)")

plt.title("Decision Tree Comparison")

plt.xticks(rotation=10)

plt.show()


# ============================================================
# BEST MODEL
# ============================================================

if gini_accuracy > entropy_accuracy:

    best_model = gini_model

    print("\nBest Model : Gini")

else:

    best_model = entropy_model

    print("\nBest Model : Entropy")


print("\nModel Training Completed Successfully!")
```

```python
# ============================================================
# DECISION TREE CLASSIFICATION PROJECT
# PART 3 - ADVANCED ANALYSIS
# ============================================================

# ===============================
# IMPORT LIBRARIES
# ===============================

from sklearn.tree import plot_tree
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV

print("="*60)
print("PART 3 STARTED")
print("="*60)


# ============================================================
# FEATURE IMPORTANCE
# ============================================================

print("\nFEATURE IMPORTANCE")

importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": best_model.feature_importances_

})

importance = importance.sort_values(

    by="Importance",

    ascending=False

)

print(importance)


# ============================================================
# FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10,5))

plt.bar(

    importance["Feature"],

    importance["Importance"]

)

plt.xticks(rotation=45)

plt.xlabel("Features")

plt.ylabel("Importance")

plt.title("Decision Tree Feature Importance")

plt.tight_layout()

plt.show()


# ============================================================
# DECISION TREE VISUALIZATION
# ============================================================

plt.figure(figsize=(22,12))

plot_tree(

    best_model,

    feature_names=X.columns,

    class_names=["No Disease","Disease"],

    filled=True,

    rounded=True,

    fontsize=9

)

plt.show()


# ============================================================
# MAX DEPTH COMPARISON
# ============================================================

print("\n" + "="*60)
print("MAX DEPTH COMPARISON")
print("="*60)

depth_list = []

accuracy_list = []

for depth in range(1,11):

    model = DecisionTreeClassifier(

        criterion="gini",

        max_depth=depth,

        random_state=42

    )

    model.fit(X_train,y_train)

    prediction = model.predict(X_test)

    acc = accuracy_score(y_test,prediction)

    depth_list.append(depth)

    accuracy_list.append(acc)

    print(

        "Depth :",depth,

        " Accuracy :",round(acc*100,2),"%"

    )


# ============================================================
# DEPTH GRAPH
# ============================================================

plt.figure(figsize=(8,5))

plt.plot(

    depth_list,

    accuracy_list,

    marker="o"

)

plt.xlabel("Tree Depth")

plt.ylabel("Accuracy")

plt.title("Tree Depth vs Accuracy")

plt.grid(True)

plt.show()


# ============================================================
# CROSS VALIDATION
# ============================================================

print("\n" + "="*60)
print("CROSS VALIDATION")
print("="*60)

cv_scores = cross_val_score(

    best_model,

    X,

    y,

    cv=5,

    scoring="accuracy"

)

print("Fold Accuracies")

print(cv_scores)

print()

print("Average Accuracy")

print(cv_scores.mean())


# ============================================================
# GRID SEARCH
# ============================================================

print("\n" + "="*60)
print("GRID SEARCH")
print("="*60)

parameters = {

    "criterion":[

        "gini",

        "entropy"

    ],

    "max_depth":[

        2,

        3,

        4,

        5,

        6,

        None

    ],

    "min_samples_split":[

        2,

        5,

        10

    ],

    "min_samples_leaf":[

        1,

        2,

        4

    ]

}

grid = GridSearchCV(

    estimator=DecisionTreeClassifier(random_state=42),

    param_grid=parameters,

    cv=5,

    scoring="accuracy",

    n_jobs=-1

)

grid.fit(X_train,y_train)

print("\nBest Parameters")

print(grid.best_params_)

print()

print("Best Cross Validation Accuracy")

print(grid.best_score_)


best_grid_model = grid.best_estimator_


# ============================================================
# TEST BEST MODEL
# ============================================================

prediction = best_grid_model.predict(X_test)

accuracy = accuracy_score(

    y_test,

    prediction

)

print()

print("Test Accuracy")

print(round(accuracy*100,2),"%")

print()

print(classification_report(

    y_test,

    prediction

))


print("="*60)
print("PART 3 COMPLETED")
print("="*60)
```

```python
# ============================================================
# DECISION TREE CLASSIFICATION PROJECT
# PART 4 - FINAL
# ============================================================

import joblib

from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve

# ============================================================
# COST COMPLEXITY PRUNING
# ============================================================

print("="*60)
print("COST COMPLEXITY PRUNING")
print("="*60)

path = DecisionTreeClassifier(
    random_state=42
).cost_complexity_pruning_path(
    X_train,
    y_train
)

alphas = path.ccp_alphas

train_scores = []
test_scores = []

for alpha in alphas:

    model = DecisionTreeClassifier(

        random_state=42,

        ccp_alpha=alpha

    )

    model.fit(X_train,y_train)

    train_scores.append(
        model.score(X_train,y_train)
    )

    test_scores.append(
        model.score(X_test,y_test)
    )

plt.figure(figsize=(8,5))

plt.plot(alphas,train_scores,label="Train")

plt.plot(alphas,test_scores,label="Test")

plt.xlabel("Alpha")

plt.ylabel("Accuracy")

plt.title("Cost Complexity Pruning")

plt.legend()

plt.show()

# ============================================================
# ROC CURVE
# ============================================================

print("\nROC AUC")

probability = best_grid_model.predict_proba(X_test)[:,1]

auc = roc_auc_score(
    y_test,
    probability
)

print("ROC AUC :",auc)

fpr,tpr,_ = roc_curve(
    y_test,
    probability
)

plt.figure(figsize=(6,6))

plt.plot(fpr,tpr)

plt.plot([0,1],[0,1])

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.grid(True)

plt.show()

# ============================================================
# SAVE MODEL
# ============================================================

joblib.dump(

    best_grid_model,

    "decision_tree_model.pkl"

)

print("\nModel Saved Successfully!")

# ============================================================
# LOAD MODEL
# ============================================================

loaded_model = joblib.load(

    "decision_tree_model.pkl"

)

print("Model Loaded Successfully!")

# ============================================================
# NEW PATIENT
# ============================================================

new_patient = pd.DataFrame([{

    "age":52,

    "sex":1,

    "cp":0,

    "trestbps":140,

    "chol":230,

    "fbs":0,

    "restecg":1,

    "thalach":150,

    "exang":0,

    "oldpeak":1.2,

    "slope":2,

    "ca":0,

    "thal":2

}])

prediction = loaded_model.predict(
    new_patient
)

probability = loaded_model.predict_proba(
    new_patient
)

print()

print("Prediction :",prediction[0])

print("Probability :",probability)

if prediction[0]==1:

    print("Heart Disease Detected")

else:

    print("No Heart Disease")

# ============================================================
# EXPORT FEATURE IMPORTANCE
# ============================================================

importance.to_csv(

    "feature_importance.csv",

    index=False

)

print()

print("Feature Importance Saved!")

print()

print("="*60)
print("PROJECT COMPLETED")
print("="*60)
```
