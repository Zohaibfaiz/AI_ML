```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================================
# STEP 1 : DATASET
# ==========================================================

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)

# ==========================================================
# STEP 2 : INITIALIZE PARAMETERS
# ==========================================================

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(X)

cost_history = []

print("="*50)
print("Gradient Descent Training Started")
print("="*50)

# ==========================================================
# STEP 3 : GRADIENT DESCENT
# ==========================================================

for epoch in range(epochs):

    # -----------------------------
    # Prediction
    # -----------------------------
    y_pred = w * X + b

    # -----------------------------
    # Error
    # -----------------------------
    error = y_pred - y

    # -----------------------------
    # Cost Function (MSE)
    # -----------------------------
    cost = np.mean(error ** 2)
    cost_history.append(cost)

    # -----------------------------
    # Gradient Calculation
    # -----------------------------
    dw = (2/n) * np.sum(error * X)
    db = (2/n) * np.sum(error)

    # -----------------------------
    # Update Parameters
    # -----------------------------
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # -----------------------------
    # Print Progress Every 100 Epochs
    # -----------------------------
    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | Cost = {cost:.6f}")

# ==========================================================
# STEP 4 : FINAL RESULTS
# ==========================================================

print("\n")
print("="*50)
print("Training Finished")
print("="*50)

print(f"\nFinal Weight (w) : {w:.4f}")
print(f"Final Bias (b)   : {b:.4f}")

print("\nFinal Linear Regression Equation")
print(f"y = {w:.4f}x + {b:.4f}")

# ==========================================================
# STEP 5 : FINAL PREDICTIONS
# ==========================================================

final_predictions = w * X + b

results = pd.DataFrame({
    "X": X,
    "Actual": y,
    "Predicted": np.round(final_predictions,2),
    "Error": np.round(y-final_predictions,2)
})

print("\nPrediction Table")
print(results)

# ==========================================================
# STEP 6 : FINAL COST
# ==========================================================

final_cost = np.mean((y-final_predictions)**2)

print("\nFinal MSE :", round(final_cost,6))

# ==========================================================
# STEP 7 : CONVERGENCE CHECK
# ==========================================================

print("\nConvergence Check")

if cost_history[-1] < cost_history[0]:
    print("Model Converged Successfully")
else:
    print("Model Did Not Converge")

# ==========================================================
# STEP 8 : COST GRAPH
# ==========================================================

plt.figure(figsize=(8,5))

plt.plot(cost_history)

plt.title("Gradient Descent Cost vs Epoch")

plt.xlabel("Epoch")

plt.ylabel("MSE Cost")

plt.grid(True)

plt.show()

# ==========================================================
# STEP 9 : REGRESSION LINE
# ==========================================================

plt.figure(figsize=(6,5))

plt.scatter(X, y, label="Actual Data")

plt.plot(X, final_predictions, label="Regression Line")

plt.xlabel("X")

plt.ylabel("Y")

plt.title("Linear Regression using Gradient Descent")

plt.legend()

plt.grid(True)

plt.show()

# ==========================================================
# STEP 10 : ACTUAL VS PREDICTED
# ==========================================================

plt.figure(figsize=(6,5))

plt.scatter(y, final_predictions)

plt.plot(
    [min(y), max(y)],
    [min(y), max(y)]
)

plt.xlabel("Actual Values")

plt.ylabel("Predicted Values")

plt.title("Actual vs Predicted")

plt.grid(True)

plt.show()

# ==========================================================
# STEP 11 : RESIDUAL ANALYSIS
# ==========================================================

residuals = y - final_predictions

plt.figure(figsize=(6,5))

plt.scatter(final_predictions, residuals)

plt.axhline(0)

plt.xlabel("Predicted Values")

plt.ylabel("Residuals")

plt.title("Residual Plot")

plt.grid(True)

plt.show()

# ==========================================================
# STEP 12 : RESIDUAL DISTRIBUTION
# ==========================================================

plt.figure(figsize=(6,5))

plt.hist(residuals, bins=5)

plt.title("Residual Distribution")

plt.xlabel("Residual")

plt.ylabel("Frequency")

plt.grid(True)

plt.show()

# ==========================================================
# STEP 13 : COST REDUCTION
# ==========================================================

print("\nFirst Cost :", round(cost_history[0],4))

print("Last Cost  :", round(cost_history[-1],8))

print("\nCost Reduced Successfully!")

print("="*50)
print("Gradient Descent Completed")
print("="*50)
```
