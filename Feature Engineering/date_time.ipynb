```python
import pandas as pd

# ==========================================
# Create Sample Dataset
# ==========================================

df = pd.DataFrame({
    "Order_Date": [
        "2024-01-05",
        "2024-02-12",
        "2024-03-25",
        "2024-04-18",
        "2024-05-30",
        "2024-06-15",
        "2024-07-20",
        "2024-08-08",
        "2024-09-10",
        "2024-10-28"
    ]
})

# ==========================================
# Convert to Datetime
# ==========================================

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# ==========================================
# Extract Date Features
# ==========================================

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()

df["Quarter"] = df["Order_Date"].dt.quarter

df["Day"] = df["Order_Date"].dt.day

df["Day_Name"] = df["Order_Date"].dt.day_name()

df["Day_of_Week"] = df["Order_Date"].dt.dayofweek
# Monday = 0
# Sunday = 6

df["Day_of_Year"] = df["Order_Date"].dt.dayofyear

df["Week_Number"] = df["Order_Date"].dt.isocalendar().week

df["Weekend"] = df["Day_of_Week"] >= 5

df["Leap_Year"] = df["Order_Date"].dt.is_leap_year

df["Days_in_Month"] = df["Order_Date"].dt.days_in_month

df["Start_of_Month"] = df["Order_Date"].dt.is_month_start

df["End_of_Month"] = df["Order_Date"].dt.is_month_end

df["Start_of_Year"] = df["Order_Date"].dt.is_year_start

df["End_of_Year"] = df["Order_Date"].dt.is_year_end

# ==========================================
# Display Dataset
# ==========================================

print("="*80)
print("Complete Date Time Analysis")
print("="*80)

print(df)

```

```python
import pandas as pd

# ==========================================
# Sample Dataset
# ==========================================

df = pd.DataFrame({
    "Date_Time": [
        "2024-01-05 08:15:30",
        "2024-02-12 13:45:10",
        "2024-03-25 18:20:55",
        "2024-04-18 23:59:59",
        "2024-05-30 00:05:15",
        "2024-06-15 11:30:45"
    ]
})

# ==========================================
# Convert to Datetime
# ==========================================

df["Date_Time"] = pd.to_datetime(df["Date_Time"])

# ==========================================
# TIME FEATURES
# ==========================================

# Hour
df["Hour"] = df["Date_Time"].dt.hour

# Minute
df["Minute"] = df["Date_Time"].dt.minute

# Second
df["Second"] = df["Date_Time"].dt.second

# Microsecond
df["Microsecond"] = df["Date_Time"].dt.microsecond

# Nanosecond
df["Nanosecond"] = df["Date_Time"].dt.nanosecond

# AM / PM
df["AM_PM"] = df["Hour"].apply(lambda x: "AM" if x < 12 else "PM")

# 12 Hour Format
df["Hour_12"] = df["Hour"] % 12
df["Hour_12"] = df["Hour_12"].replace(0, 12)

# Time Only
df["Only_Time"] = df["Date_Time"].dt.time

# Hour:Minute
df["Hour_Minute"] = df["Date_Time"].dt.strftime("%H:%M")

# Full Time
df["Full_Time"] = df["Date_Time"].dt.strftime("%H:%M:%S")

# Morning / Afternoon / Evening / Night
def period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["Time_Period"] = df["Hour"].apply(period)

# Seconds Since Midnight
df["Seconds_Since_Midnight"] = (
    df["Hour"] * 3600 +
    df["Minute"] * 60 +
    df["Second"]
)

# Minutes Since Midnight
df["Minutes_Since_Midnight"] = (
    df["Hour"] * 60 +
    df["Minute"]
)

# Difference From Previous Record
df["Time_Difference"] = df["Date_Time"].diff()

# ==========================================
# Display Result
# ==========================================

print("="*100)
print("TIME FEATURE ANALYSIS")
print("="*100)

print(df)

print("\nColumns Created:")
print(df.columns.tolist())
```
