```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 25, np.nan, 30, 35],
    "Salary": [50000, np.nan, 60000, 70000, 80000],
    "City": ["Lahore", "Karachi", "Multan", np.nan, "Islamabad"]
})

print("Missing Values")
print(df.isnull().sum())

print("\nMissing Percentage")
print((df.isnull().mean()) * 100)

# Apply CCA
df = df.dropna()

print("\nAfter CCA")
print(df)
```

```python
import pandas as pd

# ==============================
# Load Dataset
# ==============================
df = pd.read_csv("employees.csv")    # Change to your dataset path

# ==============================
# Basic Information
# ==============================
print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(f"\nRows Before CCA : {df.shape[0]}")
print(f"Columns          : {df.shape[1]}")

# ==============================
# Missing Values
# ==============================
print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())

# ==============================
# Missing Percentage
# ==============================
print("\n" + "=" * 60)
print("MISSING VALUE PERCENTAGE")
print("=" * 60)

missing_percent = (df.isnull().mean() * 100).round(2)
print(missing_percent)

# ==============================
# Apply Complete Case Analysis
# ==============================
df_cca = df.dropna()

# ==============================
# Shape Comparison
# ==============================
print("\n" + "=" * 60)
print("SHAPE COMPARISON")
print("=" * 60)

print(f"Before CCA : {df.shape}")
print(f"After CCA  : {df_cca.shape}")

# ==============================
# Data Loss
# ==============================
rows_before = df.shape[0]
rows_after = df_cca.shape[0]

rows_removed = rows_before - rows_after
loss_percent = (rows_removed / rows_before) * 100

print("\n" + "=" * 60)
print("DATA LOSS")
print("=" * 60)

print(f"Rows Removed        : {rows_removed}")
print(f"Remaining Rows      : {rows_after}")
print(f"Data Loss Percentage: {loss_percent:.2f}%")

# ==============================
# Decision
# ==============================
print("\n" + "=" * 60)
print("CCA DECISION")
print("=" * 60)

if loss_percent <= 5:
    print("✅ CCA is a GOOD choice.")
    print("Reason: Very little data is lost (<5%).")

elif loss_percent <= 10:
    print("⚠️ CCA is ACCEPTABLE.")
    print("Reason: Moderate data loss. Check whether the removed rows are important.")

elif loss_percent <= 20:
    print("⚠️ Be careful using CCA.")
    print("Reason: Significant amount of data is removed.")
    print("Consider imputation methods.")

else:
    print("❌ CCA is NOT recommended.")
    print("Reason: Too much data is removed.")
    print("Use Mean/Median/Mode/KNN/MICE imputation instead.")

# ==============================
# Missing Values After CCA
# ==============================
print("\n" + "=" * 60)
print("MISSING VALUES AFTER CCA")
print("=" * 60)

print(df_cca.isnull().sum())

# ==============================
# Preview
# ==============================
print("\n" + "=" * 60)
print("FIRST 5 ROWS AFTER CCA")
print("=" * 60)

print(df_cca.head())

# ==============================
# Final Shape
# ==============================
print("\n" + "=" * 60)
print("FINAL SHAPE")
print("=" * 60)

print(df_cca.shape)
```
